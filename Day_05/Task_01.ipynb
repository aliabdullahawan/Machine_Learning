# Day 5 Master Architecture: Logistic Regression & Decision Trees
**Objective:** Understand logistic regression as a probabilistic classification model, not just a sklearn class[cite: 2]. Then, implement the impurity metrics of Decision Trees.

```python
import numpy as np
import pandas as pd

print("--- PHASE 1: LOGISTIC REGRESSION MATH ---")

# 1. The Sigmoid Squasher
def sigmoid_by_hand(z):
    """A linear score is transformed into a probability using the sigmoid[cite: 2]."""
    return 1 / (1 + np.exp(-z))

# 2. Binary Cross-Entropy (Log Loss)
def log_loss_by_hand(y_true, y_prob):
    """Binary cross-entropy strongly penalizes confident wrong predictions[cite: 2]."""
    epsilon = 1e-15 # Prevent log(0) crash
    y_prob = np.clip(y_prob, epsilon, 1 - epsilon)
    return -np.mean(y_true * np.log(y_prob) + (1 - y_true) * np.log(1 - y_prob))

# We successfully implement sigmoid and log loss with NumPy[cite: 2].
print("Sigmoid of Log-Odds 1.38:", round(sigmoid_by_hand(1.38), 2))
```

# Cell 3: Python Code (PHASE 2: Decision Tree Impurity By Hand)

```python
print("\n--- PHASE 2: DECISION TREE CLASSIFICATION MATH ---")

# 1. Gini Impurity
def binary_gini(p0, p1):
    """For the binary case: Gini = 1 - (p0^2 + p1^2)[cite: 2]."""
    return 1 - (p0**2 + p1**2)

# 2. Entropy 
def binary_entropy(p0, p1):
    """For binary classification: H = -p0 log2(p0) - p1 log2(p1)[cite: 2]."""
    epsilon = 1e-15
    p0, p1 = max(p0, epsilon), max(p1, epsilon)
    return -(p0 * np.log2(p0)) - (p1 * np.log2(p1))

# 3. Splitting & Information Gain
def evaluate_split(n, h_parent, n_L, h_L, n_R, h_R):
    """
    Weighted Gini after splitting: Gini_split = (n_L / n) * Gini_L + (n_R / n) * Gini_R[cite: 2].
    The tree wants a lower resulting impurity[cite: 2].
    
    For two children: IG = H(parent) - [ (n_L/n)*H(L) + (n_R/n)*H(R) ][cite: 2].
    The tree wants to maximize information gain[cite: 2].
    """
    weighted_child_impurity = (n_L / n) * h_L + (n_R / n) * h_R
    information_gain = h_parent - weighted_child_impurity
    return weighted_child_impurity, information_gain

# Test a 50/50 chaotic node
print("Gini of a 50/50 split:", binary_gini(0.5, 0.5))
print("Entropy of a 50/50 split:", binary_entropy(0.5, 0.5))
```

# Cell 4: Python Code (PHASE 3: Regression Tree Variance By Hand)

```python
print("\n--- PHASE 3: REGRESSION TREE MATH ---")

def node_variance(y_node):
    """
    Regression Tree — MSE: For a node, MSE = (1/n) * sum(y_i - y_bar)^2[cite: 2].
    Regression Tree — variance: Var(Y) = (1/n) * sum(y_i - y_bar)^2[cite: 2].
    """
    n = len(y_node)
    if n == 0: return 0
    y_bar = np.mean(y_node)
    return np.sum((y_node - y_bar)**2) / n

# A regression tree searches for splits that reduce the prediction error/variance inside the resulting nodes[cite: 2].
good_node = np.array([10, 11, 12]) 
bad_node = np.array([2, 11, 25])
print("Variance of tightly clustered node:", round(node_variance(good_node), 2))
print("Variance of messy, scattered node:", round(node_variance(bad_node), 2))
```
