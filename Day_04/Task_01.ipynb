# Phase 4: Demand Prediction & Regularization Architecture
**Deliverables included:** A Demand prediction project, a Complexity curve, a Regularization experiment table, and an Explanation of the bias/variance tradeoff[cite: 3].

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

```

```python
# 1. Dataset: Demand Prediction Project (e.g., Active Users vs Server Demand)[cite: 3]
np.random.seed(42)
n_samples = 100
X = np.sort(np.random.uniform(0, 10, n_samples)).reshape(-1, 1)

# True relationship is a complex curve (Sine wave + exponential + noise)
y = np.sin(X).ravel() * 10 + (X.ravel() ** 2) + np.random.normal(0, 5, n_samples)

# The Hard Way: Manual Feature Engineering
X_squared = X ** 2
X_cubed = X ** 3
X_quad = X ** 4
X_poly = np.column_stack((X, X_squared, X_cubed, X_quad))

# 2. Strict Split
X_train, X_test, y_train, y_test = train_test_split(X_poly, y, test_size=0.2, random_state=42)


# 3. Baseline Linear Model (Degree 1)[cite: 3]
baseline_model = LinearRegression()
baseline_model.fit(X_train, y_train)
baseline_pred = baseline_model.predict(X_test)
baseline_mse = mean_squared_error(y_test, baseline_pred)
baseline_rmse = np.sqrt(mean_squared_error(y_test, baseline_pred))
baseline_mae = mean_absolute_error(y_test, baseline_pred)
baseline_r2 = r2_score(y_test, baseline_pred)
print(f"Baseline Linear MSE: {baseline_mse:.2f}")
print(f"Baseline Linear RMSE: {baseline_rmse:.2f}")
print(f"Baseline Linear MAE: {baseline_mae:.2f}")
print(f"Baseline Linear R2_Score: {baseline_r2:.2f}")
```

## Cell 3: Python Code (The Complexity Curve Experiment)

```python
# Experiment: Increase polynomial degree one step at a time and track training vs validation error[cite: 3]

degrees = range(1, 15)
train_errors = []
val_errors = []

for d in degrees:
    # 1. Generate Polynomial Features[cite: 3]
    poly = PolynomialFeatures(degree=d, include_bias=False)
    X_train_poly = poly.fit_transform(X_train)
    X_test_poly = poly.transform(X_test)
    
    # 2. Train Model
    model = LinearRegression()
    model.fit(X_train_poly, y_train)
    
    # 3. Track Errors
    train_pred = model.predict(X_train_poly)
    test_pred = model.predict(X_test_poly)
    
    rmse_train = np.sqrt(mean_squared_error(y_train, train_pred))
    rmse_test = np.sqrt(mean_squared_error(y_test, test_pred))
    
    print(f"RMSE Train: {rmse_train}")
    print(f"RMSE Test: {rmse_test}")
    
    train_errors.append(rmse_train)
    val_errors.append(rmse_test)

# 4. Validate Complexity (Plotting the Complexity Curve)[cite: 3]
plt.figure(figsize=(10, 5))
plt.plot(degrees, train_errors, label='Training Error (Blue)', marker='o')
plt.plot(degrees, val_errors, label='Validation Error (Red)', marker='s', color='red')
plt.title("Model Complexity Curve (Bias/Variance Tradeoff)")
plt.xlabel("Polynomial Degree")
plt.ylabel("RMSE")
plt.ylim(0, 20)
plt.legend()
plt.grid(True)
plt.show()

print("Look at the graph: Around Degree 3-5, validation error drops (Sweet Spot).")
print("At Degree 10+, Training Error drops to near zero, but Validation Error EXPLODES (Severe Overfitting)[cite: 3].")
```

## Cell 4: Python Code (The Regularization Experiment)

```python
# Experiment: Vary regularization strength and compare coefficient magnitudes[cite: 3]
# We will intentionally use a severely overfitted Degree 12 polynomial.

poly = PolynomialFeatures(degree=12, include_bias=False)
X_train_p12 = poly.fit_transform(X_train)
X_test_p12 = poly.transform(X_test)

# CRITICAL: Feature scaling is particularly important when regularization penalties act directly on coefficient magnitudes[cite: 3].
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_p12)
X_test_scaled = scaler.transform(X_test_p12)

# Test different regularization strengths (alphas)[cite: 3]
alphas = [0.001, 1.0, 100.0]
results = []

for a in alphas:
    # Train Ridge
    ridge = Ridge(alpha=a)
    ridge.fit(X_train_scaled, y_train)
    r_pred = ridge.predict(X_test_scaled)
    r_rmse = np.sqrt(mean_squared_error(y_test, r_pred))
    
    # Train Lasso
    lasso = Lasso(alpha=a, max_iter=10000)
    lasso.fit(X_train_scaled, y_train)
    l_pred = lasso.predict(X_test_scaled)
    l_rmse = np.sqrt(mean_squared_error(y_test, l_pred))
    
    # Compare coefficient magnitudes[cite: 3]
    r_weight_sum = np.sum(np.abs(ridge.coef_))
    # Count how many coefficients Lasso set to EXACTLY zero[cite: 3]
    l_zeros = np.sum(lasso.coef_ == 0)
    
    results.append({
        "Alpha": a,
        "Ridge RMSE": round(r_rmse, 2),
        "Ridge Total Weight Mass": round(r_weight_sum, 2),
        "Lasso RMSE": round(l_rmse, 2),
        "Lasso Deleted Features (Zeros)": l_zeros
    })

# Output the Regularization experiment table[cite: 3]
print("\n--- REGULARIZATION EXPERIMENT TABLE ---")
print(pd.DataFrame(results).to_string(index=False))
```

### Final Deliverable: Explanation of Bias/Variance Tradeoff[cite: 3]
The Bias/Variance tradeoff is the ultimate balancing act in machine learning architecture:
*   **High Bias (Underfitting):** The model is too simple (e.g., a straight line on curved demand data). It makes massive assumptions and fails to capture the true pattern. Training error is high; validation error is high.
*   **High Variance (Overfitting):** The model is too complex (e.g., a Degree 12 polynomial). It memorizes the random noise in the training set. Training error is zero; validation error explodes to infinity.
*   **The Tradeoff Solution:** We generate polynomial features to cure High Bias, and then we apply Ridge/Lasso Regularization to cure the resulting High Variance. We intentionally trade a little training fit for better generalization[cite: 3] in production.
