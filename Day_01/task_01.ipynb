# Day 1: The ML Vocabulary & Mental Models
**Objective:** Establish the mathematical and architectural language required to build, train, and deploy ML systems.

### 1. The Global ML Workflow Context
Machine learning is not pattern memorization; it is **parameter estimation** and **generalization**. Everything we define here maps to our 19-step deployment pipeline.

### 2. Learning Paradigms
* **Supervised Learning:** We are provided with inputs $X$ and a known target $y$. 
  * *Regression:* Predicts a continuous numerical quantity.
  * *Classification:* Predicts a discrete class or a probability distribution over classes.
* **Unsupervised Learning:** We are provided with inputs $X$ but no target $y$. The system infers the underlying structure independently.

### 3. The Data Layer
* **Observation:** A single, distinct instance of data. If you are designing a database schema, an observation is exactly one row in your SQL table.
* **Features ($X$):** The independent variables or inputs used to make a decision. 
* **Target ($y$):** The exact dependent variable we are attempting to predict.

```python
import numpy as np
import pandas as pd

# Simulating Step 4: Acquire/query data
# Here, each dictionary represents one 'Observation'
data = [
    {"feature_1_price": 29.99, "feature_2_downloads": 500, "target_future_sales": 150},
    {"feature_1_price": 59.99, "feature_2_downloads": 1200, "target_future_sales": 400},
    {"feature_1_price": 9.99, "feature_2_downloads": 300, "target_future_sales": 80}
]

df = pd.DataFrame(data)

# Separating Features (X) and Target (y)
X = df[["feature_1_price", "feature_2_downloads"]].values
y = df["target_future_sales"].values

print(f"Features Shape: {X.shape} (3 observations, 2 features)")
print(f"Target Shape: {y.shape} (3 observations, 1 target)")
```

### 4. The Modeling Layer
* **Estimator:** The algorithmic function used to map features to the target. 
* **Parameter ($\theta$ or $w$):** The internal variables the estimator learns purely from the training data. The model updates these automatically during optimization.
* **Hyperparameter:** The configuration variables set *before* training begins (e.g., learning rate $\alpha$). 
* **Loss:** A mathematical penalty representing the error of a *single* prediction compared to the actual target.
* **Objective:** The overarching mathematical function the model minimizes or maximizes across the *entire* dataset.

### 5. The Lifecycle & Evaluation Layer
* **Baseline:** The simplest meaningful reference model. If our complex ML model cannot beat a simple baseline (like predicting the mean every time), it is useless.
* **Generalization:** The model's performance on unseen data from the intended deployment distribution. 
* **Artifact:** The final, serialized file containing the trained model and its established parameters. Think of this like compiling your Python scripts into a final, deployable desktop executable.
* **Inference:** The act of passing live, unseen data into your loaded artifact to generate a prediction in production.
  * *Training:* historical data -> features -> fit -> validate -> select -> save artifact
  * *Inference:* request -> validate -> same feature transformations -> load artifact -> predict -> response
